## Dimensions in array

````python
# 0-Dimension
a = np.array(42)
# 1-Dimension
b = np.array([1,2,3,4,5])
# 2-Dimension
c = np.array([[1,2,3,4], [5,6,7,8]])
# 3-Dimension
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
print('\n')

print("Zero dimension : \n", a)
print('\nOne Dimension:\n',b)
print('\nTwo Dimension:\n',c)
print('\nThree Dimension:\n',d)
````

## Matplotlib

````python
# Import library
import matplotlib.pyplot as plt
````

````python
# Basic plot
x = [1,2,3]
y = [2,4,6]

plt.plot(x,y)
plt.show()
````

````python
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Basic Plot')
plt.show()
````

````python
# We can plot more than one in the same plot
z = np.arange(1, 3.1, 0.1)

plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Basic Plot')
plt.plot(z,z**2)
plt.show()
````

````python
# We can use lable for plot meaning
z = np.arange(1, 3.1, 0.1)

plt.plot(x,y, label='y = 2x')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Basic Plot')
plt.plot(z,z**2, label='y = x^2')
plt.legend()
plt.show()
#
````

## Seaborn

````python
# Import library
import seaborn as sns
````

````python
# Set seaborn theme
sns.set_theme(style="whitegrid")

# Defien data
x = np.arange(1, 3.1, 0.1)
y = 2 * x
z = x
z_squared = z ** 2

# Plot the data
plt.figure(figsize=(10,6))
sns.lineplot(x=x, y=y, label='y = 2x', marker="o", linewidth=2)
sns.lineplot(x=z, y=z_squared, label="z²", marker="s", linewidth=2)

# Add labels and titles
plt.xlabel("X-axis", fontsize=12)
plt.ylabel("Y-axis", fontsize=12)
plt.title("Basic Graph using Seaborn", fontsize=14)
plt.legend(title="Functions")
plt.tight_layout()
plt.show()
````

## Plotline
Plotline is a Python implementation of the Grammar of Graphics - a powerful and flexible system for creating statistical graphics based on layered principles originally developed in ggplot2 for R

````python
import numpy as np
import pandas as pd
from IPython.display import display

from plotnine import (
    ggplot, aes, geom_line, geom_point,
    labs, theme_minimal, theme, element_text, scale_color_manual
)

# Prepare data
x = np.arange(1, 3.1, 0.1)
df = pd.DataFrame({
    "x": list(x) * 2,
    "y": list(2 * x) + list(x ** 2),
    "Function": ["2x"] * len(x) + ["z²"] * len(x)
})

# Build the plot using grammar of graphics
plot = (
    ggplot(df, aes(x="x", y="y", color="Function", shape="Function")) +  # Aesthetics
    geom_line(size=1.5) +                                                # Geometry: lines
    geom_point(size=2.5) +                                               # Geometry: points
    scale_color_manual(values={"2x": "#1f77b4", "z²": "#ff7f0e"}) +      # Custom colors
    labs(
        title="Graph of Functions 2x and z²",
        x="X-axis",
        y="Y-axis",
        color="Function",
        shape="Function"
    ) +
    theme_minimal() +
    theme(
        axis_title=element_text(size=12),
        plot_title=element_text(size=14, weight='bold', ha='center'),
        legend_title=element_text(size=10),
        legend_text=element_text(size=9)
    )
)

print(plot)

# At the end of your code
display(plot.draw())
````
